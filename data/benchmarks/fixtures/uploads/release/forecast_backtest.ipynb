# Daily demand backtest
Twelve daily observations, already sorted by date. No missing days.

```python
import numpy as np
from sklearn.model_selection import TimeSeriesSplit
values = np.arange(12)
cv = TimeSeriesSplit(n_splits=3, test_size=2, gap=1)
folds = list(cv.split(values))
```

## Interpretation boundary
These are index splits, not measured forecast scores. No estimator was fitted and no accuracy score is recorded.
