# Subscription snapshots
Rows are snapshots, so an account can appear in more than one month.

```python
import pandas as pd
january = pd.DataFrame({'account': ['A', 'B'], 'seats': [3, 8]}, index=[0, 1])
february = pd.DataFrame({'account': ['A', 'C'], 'seats': [4, 2]}, index=[0, 1])
stacked = pd.concat([january, february], keys=['Jan', 'Feb'])
flat = pd.concat([january, february], ignore_index=True)
```
